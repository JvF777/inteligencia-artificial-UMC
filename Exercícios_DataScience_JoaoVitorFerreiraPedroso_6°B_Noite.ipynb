```python
#Ex-19 NUMPY
import numpy as np

a = np.random.randint(1, 50, 20)
b = np.random.randint(1, 50, 20)

produto = np.dot(a, b)
tamanho_a = np.sqrt(np.sum(np.multiply(a, a)))
tamanho_b = np.sqrt(np.sum(np.multiply(b, b)))
cosseno = produto / (tamanho_a * tamanho_b)

print("Produto escalar:", produto)
print("Cosseno do ângulo:", cosseno)
print("Ângulo em graus:", np.arccos(cosseno) * 180 / np.pi)

#o produto escalar normalizado é o cosseno do ângulo entre os vetores.
#o tamanho de cada vetor sai multiplicando ele por si mesmo, somando e tirando a raiz.
#o np.arccos devolve o ângulo em radianos, então multiplico por 180 e divido por pi.
#np.pin para pegar o PI
```

```python
#Ex-1 SCIPY

import numpy as np
from scipy.integrate import quad

def f(x):
    return x**2 * np.sin(x)

resultado, erro = quad(f, 0, np.pi)
print("Integral:", resultado)
print("Erro estimado:", erro)

#o quad recebe a função como argumento, por isso defino ela antes.
#ele devolve dois valores: o resultado e o erro estimado.
```

```python
#Ex-2 SCIPY

import numpy as np
from scipy.integrate import solve_ivp

def eq_diferencial(x, y):
    return x - y

solucao = solve_ivp(eq_diferencial, [0, 10], [1], t_eval=np.linspace(0, 10, 11))
print("x:", solucao.t)
print("y:", solucao.y[0])

#o [0, 10] é o intervalo e o [1] é a condição inicial y(0) = 1.
#o t_eval define em quais pontos quero ver a resposta.
#solucao.t traz os x e solucao.y[0] traz os y correspondentes.
```

```python
#Ex-3 SCIPY

from scipy.optimize import root

def f(x):
    return x**4 - 3*x**3 + 2

for chute in [0.5, 1, 3]:
    solucao = root(f, chute)
    print("A partir", chute, "-> raiz:", solucao.x[0])

#o root acha a raiz mais próxima do chute inicial, uma por vez.
#testo vários chutes porque a função é de grau 4 e tem mais de uma raiz.
```

```python
#Ex-4 SCIPY

from scipy.optimize import minimize

def f(x):
    return -x**2 + 4*x + 1

def f_negativa(x):
    return -f(x)

resultado = minimize(f_negativa, 0)
print("Máximo em x =", resultado.x[0])
print("Valor máximo:", f(resultado.x[0]))

#o SciPy só minimiza, então inverto o sinal da função.
#o ponto mais baixo da função invertida é o mais alto da original.
#o 0 no minimize é o chute inicial.
```

```python
#Ex-9 SCIPY

import numpy as np
from scipy.optimize import curve_fit

x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 7, 8])

def parabola(x, a, b, c):
    return a * x**2 + b * x + c

parametros, _ = curve_fit(parabola, x, y)
print("Parametros ajustados [a, b, c]:", parametros)


#o curve_fit procura os valores de a, b e c que aproximam a curva dos pontos.
#a função modelo recebe o x primeiro e depois os parâmetros a descobrir.
```

```python
#Ex-10 SCIPY


from scipy.integrate import quad

def integral_interna(x):
    def f(y):
        return x * y + x**2
    resultado, _ = quad(f, 0, 1)
    return resultado

resultado, erro = quad(integral_interna, 0, 2)
print("Integral dupla:", resultado)

#primeiro integro em y de 0 a 1 tratando o x como fixo.
#o resultado disso vira a função que integro em x de 0 a 2.
```

```python
#Ex-16 SCIPY


import numpy as np
from scipy.integrate import solve_ivp

def sistema(x, u):
    return [u[1], -u[0]]

solucao = solve_ivp(sistema, [0, 10], [1, 0], t_eval=np.linspace(0, 10, 11))
print("y calculado:", solucao.y[0])
print("y exato (cos x):", np.cos(solucao.t))


#quebro a equação de segunda ordem em duas de primeira.
#chamo u[0] de y e u[1] de y', e como y'' = -y a derivada de u[1] é -u[0].
#a condição inicial [1, 0] junta y(0)=1 e y'(0)=0.
```

```python
#Ex-19 SCIPY


import numpy as np
from scipy.fft import fft

t = np.linspace(0, 1, 256)
sinal = np.sin(2 * np.pi * 5 * t) + 0.5 * np.cos(2 * np.pi * 20 * t)

transformada = fft(sinal)
magnitudes = np.sqrt(np.multiply(transformada.real, transformada.real) +
                     np.multiply(transformada.imag, transformada.imag))

for i in range(30):
    if magnitudes[i] > 10:
        print("Pico na frequência", i, "Hz - magnitude:", magnitudes[i])


#monto o sinal com uma frequência de 5 Hz e outra de 20 Hz.
#o resultado do fft é complexo, então calculo a magnitude com Pitágoras.
#o if filtra os valores altos, que são os picos das frequências do sinal.
```

```python
#Ex-11 PANDAS

import numpy as np
import pandas as pd

dados = {'Produto': ['Cimento', 'Tijolo', 'Areia', 'Telha', 'Tinta'],
         'Preco': [np.nan, 1.2, 90.0, np.nan, 28.0]}
df = pd.DataFrame(dados)

df['Preco'] = df['Preco'].fillna(df['Preco'].median())
print(df)


#o fillna preenche as casas vazias em vez de apagar a linha.
#calculo a mediana da própria coluna e uso como valor de preenchimento.
#preciso atribuir de volta, senão a alteração não fica salv
```

```python
#Ex-12 PANDAS


import pandas as pd

dados = {'ID': [1, 2, 3, 4],
         'Produto': ['Ventilador', 'Liquidificador', 'Ventilador', 'Batedeira'],
         'Preco': [180, 250, 320, 250]}
df = pd.DataFrame(dados)
print(df, "\n")

for coluna in df.columns:
    if df[coluna].duplicated().any():
        print(coluna, ": tem duplicados")
    else:
        print(coluna, ": não tem duplicados")

#o df.columns devolve os nomes das colunas para o for percorrer.
#o duplicated marca como verdadeiro os valores que já apareceram antes.
#o any resume tudo em uma resposta só.
```

```python
#Ex-8 MATPLOTLIB


import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y, label='sin(x)')
plt.fill_between(x, y - 0.2, y + 0.2, alpha=0.2)
plt.title("Seno com Intervalo de Confiança")
plt.legend()
plt.show()


#o fill_between pinta a área entre dois limites, um abaixo e outro acima da curva.
#uso 0.2 para cima e para baixo, criando a margem em torno do seno.
```

```python
#Ex-11 MATPLOTLIB

import matplotlib.pyplot as plt
import numpy as np

categorias = ['A', 'B', 'C']
grupo1 = np.array([3, 5, 2])
grupo2 = np.array([2, 4, 6])
grupo3 = np.array([4, 1, 3])

plt.bar(categorias, grupo1, label='Grupo 1')
plt.bar(categorias, grupo2, bottom=grupo1, label='Grupo 2')
plt.bar(categorias, grupo3, bottom=grupo1 + grupo2, label='Grupo 3')
plt.title("Barras Empilhadas")
plt.xlabel("Categorias")
plt.ylabel("Valores")
plt.legend()
plt.show()


#o bottom diz onde cada barra começa, e é o que faz o empilhamento.
#o terceiro grupo começa em cima da soma dos dois anteriores.
#uso array do NumPy para poder somar grupo1 + grupo2 elemento por elemento.
```

```python
#Ex-18 MATPLOTLIB


import matplotlib.pyplot as plt

meses = ['Jan', 'Fev', 'Mar', 'Abr', 'Mai', 'Jun']
valores = [5, -3, 8, -6, 2, -1]

cores = []
for valor in valores:
    if valor >= 0:
        cores.append('green')
    else:
        cores.append('red')

plt.bar(meses, valores, color=cores)
plt.title("Positivos (verde) e Negativos (vermelho)")
plt.xlabel("Mês")
plt.ylabel("Valor")
plt.show()


#monto a lista de cores com um laço, decidindo pela sinal de cada valor.
#a lista fica na mesma ordem dos valores, então cada barra pega a cor certa.
```
