```python
#ex-22

alfabeto = "abcdefghijklmnopqrstuvwxyz"

def criptografar(msg, desloc):
    resultado = ""
    for letra in msg:
        if letra in alfabeto:
            pos = alfabeto.index(letra)
            nova_pos = (pos + desloc) % 26
            resultado = resultado + alfabeto[nova_pos]
        else:
            resultado = resultado + letra

    return resultado


mensagem = input("Digite a mensagem: ").lower()
deslocamento = int(input("Digite o deslocamento: "))

print(f"Mensagem criptografada: {criptografar(mensagem, deslocamento)}")

#uso def para criar função dentro do codigo.
#criei uma string com o alfabeto e uso ela como referência para achar as posições.
#o index() devolve em que posição a letra está e depois somo o deslocamento.
#o % 26 faz o alfabeto dar a volta: o z com deslocamento 3 volta para o começo e vira c.
#se o caractere não estiver no alfabeto, como espaço ou vírgula, copio ele sem mudar nada.
```

```python
#ex-23


alfabeto = "abcdefghijklmnopqrstuvwxyz"

def descriptografar(msg, desloc):
    resultado = ""

    for letra in msg:
        if letra in alfabeto:
            pos = alfabeto.index(letra)
            nova_pos = (pos - desloc) % 26
            resultado = resultado + alfabeto[nova_pos]
        else:
            resultado = resultado + letra

    return resultado


mensagem = input("Digite a mensagem cifrada: ").lower()
deslocamento = int(input("Digite o deslocamento usado: "))

print(f"Mensagem original: {descriptografar(mensagem, deslocamento)}")

#a única diferença para o Ex-22 é o sinal, aqui subtraio o deslocamento em vez de somar.
```

```python
#ex-24

cpf = input("Digite o CPF (só números): ")

valido = True

if len(cpf) != 11:
    valido = False
elif cpf == cpf[0] * 11:
    valido = False
else:
    soma1 = 0
    peso = 10
    for i in range(9):
        soma1 = soma1 + int(cpf[i]) * peso
        peso = peso - 1

    digito1 = 11 - (soma1 % 11)
    if digito1 > 9:
        digito1 = 0

    soma2 = 0
    peso = 11
    for i in range(10):
        soma2 = soma2 + int(cpf[i]) * peso
        peso = peso - 1

    digito2 = 11 - (soma2 % 11)
    if digito2 > 9:
        digito2 = 0

    if int(cpf[9]) != digito1:
        valido = False
    if int(cpf[10]) != digito2:
        valido = False

if valido == True:
    print("CPF válido")
else:
    print("CPF inválido")

#uso uma variável peso que começa em 10 e vai diminuindo de 1 em 1 a cada volta do laço.
#multiplico cada número pelo peso, somo tudo e tiro o resto da divisão por 11.
#depois subtraio esse resto de 11 e, se passar de 9, o dígito vira 0.
#o segundo dígito segue a mesma regra, mas começa com peso 11 e usa 10 números.
```

```python
#ex-30

import random

palavras = ["java", "corinthians", "processador",]
palavra = random.choice(palavras)

letras_certas = []
erros = 0

print("Jogo da Forca!")
print(f"A palavra tem {len(palavra)} letras. Você pode errar 6 vezes.")

while erros < 6:
    exibicao = ""
    for letra in palavra:
        if letra in letras_certas:
            exibicao = exibicao + letra
        else:
            exibicao = exibicao + "_"

    if exibicao == palavra:
        print(f"Parabéns! A palavra era {palavra}")
        break

    print(f"Palavra: {exibicao}")
    print(f"Erros: {erros} de 6")

    tentativa = input("Digite uma letra: ").lower()

    if tentativa in palavra:
        letras_certas.append(tentativa)
        print("Acertou a letra!")
    else:
        erros = erros + 1
        print("Letra errada!")

if erros == 6:
    print(f"Você perdeu! A palavra era {palavra}")

#uso o choice() para sortear uma palavra da lista.
#guardo as letras acertadas em uma lista.
#o operador in verifica se a letra existe na palavra, então não preciso percorrer letra por letra.
#a montagem da exibição fica no começo do laço para checar a vitória antes de pedir nova letra.
#o while roda enquanto os erros forem menores que 6, e o break sai quando a palavra é completada.
```

```python
#Ex-40

compras = []
opcao = ""

while opcao != "4":
    print()
    print("1 - Adicionar item")
    print("2 - Remover item")
    print("3 - Listar itens")
    print("4 - Sair")

    opcao = input("Escolha uma opção: ")

    match opcao:
        case "1":
            item = input("Digite o item que quer adicionar: ")
            compras.append(item)
            print(f"{item} foi adicionado!")

        case "2":
            item = input("Digite o item que quer remover: ")
            if item in compras:
                compras.remove(item)
                print(f"{item} foi removido!")
            else:
                print("Esse item não está na lista!")

        case "3":
            if len(compras) == 0:
                print("A lista está vazia!")
            else:
                print("Lista:")
                for item in compras:
                    print("-", item)

        case "4":
            print("Saida feita!")

        case _:
            print("Opção inválida!")

#uso uma lista vazia para guardar os itens e um while que roda até o usuário escolher a opção 4.
#estou acostumado usar switch no java, porém no Python não tem, pesquisei e achei o comando match(), aonde é mais aparecido.
#o case _ funciona como o else e pega qualquer opção que não seja de 1 a 4.
#a opção é lida como texto, por isso uso "".
#o append() adiciona o item no final da lista e o remove() tira o item pelo nome.
#antes de remover confiro com o in se o item existe, senão o remove() daria erro.
```

```python
#ex-41

numeros = [1, 13, 7, 4, 5, 5, 0, 3, 14]
sem_duplicatas = []

for n in numeros:
    if n in sem_duplicatas:
        print(f"{n} já está na lista, ignorando")
    else:
        sem_duplicatas.append(n)

print(f"Lista original: {numeros}")
print(f"Sem duplicatas: {sem_duplicatas}")

#crio uma lista vazia e percorro a original adicionando os números nela.
#o in verifica se o número já foi adicionado antes.
#o else só adiciona quando é a primeira vez que o número aparece.
```

```python
#ex-42

n = int(input("Digite até qual número quer os primos: "))

for num in range(2, n + 1):
    primo = True

    for i in range(2, num):
        if num % i == 0:
            primo = False
            break

    if primo == True:
        print(num)

#uso a mesma lógica do Ex-18, mas agora testando vários números em vez de um só.
#por isso preciso de dois laços: o de fora percorre os candidatos e o de dentro procura divisores.
# o range() começa no 2, porque 0 e 1 não sao primos.
#a variável primo precisa voltar a ser True a cada número novo, por isso ela fica dentro do laço de fora.
#o break sai assim que acha o primeiro divisor.
```

```python
#ex-43

print("1 - Metros para centímetros")
print("2 - Metros para quilômetros")
print("3 - Centímetros para metros")
print("4 - Quilômetros para metros")

opcao = input("Escolha a conversão: ")
valor = float(input("Digite o valor: "))

match opcao:
    case "1":
        print(f"{valor} m = {valor * 100} cm")
    case "2":
        print(f"{valor} m = {valor / 1000} km")
    case "3":
        print(f"{valor} cm = {valor / 100} m")
    case "4":
        print(f"{valor} km = {valor * 1000} m")
    case _:
        print("Opção inválida!")

#uso o match/case no menu, mesma estrutura que usei na lista de compras.
#uso float para aceitar valores com casas decimais.
#multiplico para unidade menor
#divido para unidade maior
```

```python
#ex-45

numeros = [14, 3, 27, 40, 11, 6, 33, 58, 25]

pares = []
impares = []

for n in numeros:
    if n % 2 == 0:
        pares.append(n)
    else:
        impares.append(n)

print(f"Lista original: {numeros}")
print(f"Pares: {pares}")
print(f"Ímpares: {impares}")

#uso o % para saber se o número é par.
#crio listas vazias para cada caso, tanto quanto impares e pares.
#o append(), adiciona o numero em cada lista vazia
```

```python
#ex-46

numeros = [14, 3, 27, 40, 11, 6, 33, 58, 25]

soma = 0
for n in numeros:
    soma = soma + n

media = soma / len(numeros)

soma_quadrados = 0
for n in numeros:
    dif = n - media
    soma_quadrados = soma_quadrados + dif * dif

variancia = soma_quadrados / len(numeros)
desvio = variancia ** 0.5

print(f"Média: {media}")
print(f"Desvio padrão: {desvio}")

#faço a mesma logica da media do ex-15.
#crio a varivel soma_quadrados, para somar dentro do for().
#calculo a diferença de cada número até a média e multiplico ela por ela mesma.
#uso ** 0.5 para tirar a raiz quadrada.
```

```python
#ex-47

from collections import Counter

numeros = [3, 7, 3, 2, 9, 7, 1, 2, 9, 3]

contagem = Counter(numeros)

for numero, quantidade in contagem.items():
    print(f"O número {numero} aparece {quantidade} vezes")

#o Counter recebe a lista e já devolve um map com cada número e quantas vezes ele aparece.
#usei ele no Ex-12 para contar caracteres, aqui funciona igual mas com números.
#o items() devolve a chave e o valor juntos, por isso uso duas variáveis no for.
```

```python
#ex-48

lista = [42, 8, 15, 4, 23, 16, 30, 11]
ordenada = []

for n in lista:
    pos = 0
    while pos < len(ordenada):
        if ordenada[pos] < n:
            pos = pos + 1
        else:
            break
    ordenada.insert(pos, n)

print("Lista original:", lista)
print("Lista nova:", ordenada)

#aqui monto uma lista nova, então a original não é alterada e dá para comparar as duas no final.
#para cada número percorro a lista ordenada procurando a primeira posição onde ele deve entrar.
#o while avança enquanto encontra números menores e o break para quando acha um maior.
#o insert() coloca o número naquela posição e empurra os que vêm depois para a frente.
#o pos < len(ordenada) impede que eu passe do fim da lista.
```

```python
#ex-49

numero = input("Digite um número inteiro: ")

soma = sum(map(int, numero))

print(f"A soma dos dígitos é: {soma}")

#deixo o numero como texto, para ler digito por digito.
#o map(int, numero) aplica a função int em cada caractere, convertendo todos de uma vez.
#o sum() soma os valores convertidos.
```
