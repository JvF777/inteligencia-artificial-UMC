## 11. Dados Ausentes

```python
df_remocao = df.dropna()

print("ESTRATÉGIA A, remoção de observações incompletas\n")
print(f"  Registros antes: {df.shape[0]}")
print(f"  Registros depois: {df_remocao.shape[0]}")
perda = (df.shape[0] - df_remocao.shape[0]) / df.shape[0] * 100
print(f"  Perda amostral: {perda:.1f}%")
```

```python
df_tratado = df.copy()

print("ESTRATÉGIA B, imputação pela mediana\n")
for col in df_tratado.columns:
    if df_tratado[col].isnull().sum() > 0:
        mediana = df_tratado[col].median()
        n = df_tratado[col].isnull().sum()
        df_tratado[col] = df_tratado[col].fillna(mediana)
        print(f"  {col}: {n} valores imputados com mediana {mediana:.2f}")

print(f"\n  Valores ausentes restantes: {df_tratado.isnull().sum().sum()}")
print(f"  Registros preservados: {df_tratado.shape[0]}, 100% da base")
```

## 14. Separação entre atributos e variável alvo

```python
X = df_tratado.drop(columns=["Potability"])
y = df_tratado["Potability"]

print(f"Matriz de atributos X: {X.shape}")
print(f"Vetor alvo y: {y.shape}")
print(f"\nAtributos utilizados:\n{list(X.columns)}")
```

## 15. Divisão em conjuntos de treino e teste

```python
X_treino, X_teste, y_treino, y_teste = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("DIVISÃO DA BASE")
print("=" * 50)
print(f"Treino: {X_treino.shape[0]} registros, {X_treino.shape[0]/len(X)*100:.1f}%")
print(f"Teste:  {X_teste.shape[0]} registros, {X_teste.shape[0]/len(X)*100:.1f}%")
print()
print("Proporção de classes preservada:")
print(f"  Base original: {(y.value_counts(normalize=True)*100).round(2).to_dict()}")
print(f"  Treino:        {(y_treino.value_counts(normalize=True)*100).round(2).to_dict()}")
print(f"  Teste:         {(y_teste.value_counts(normalize=True)*100).round(2).to_dict()}")
```

### Padronização

```python
scaler = StandardScaler()

X_treino_norm = scaler.fit_transform(X_treino)
X_teste_norm = scaler.transform(X_teste)

print("Padronização aplicada.\n")
comparacao = pd.DataFrame({
    "media_antes": X_treino.mean().round(2),
    "desvio_antes": X_treino.std().round(2),
    "media_depois": X_treino_norm.mean(axis=0).round(4),
    "desvio_depois": X_treino_norm.std(axis=0).round(4)
})
comparacao
```

## 16. Modelo de referência

```python
svm_base = SVC(random_state=42)
svm_base.fit(X_treino_norm, y_treino)

y_pred_base = svm_base.predict(X_teste_norm)
acc_base = accuracy_score(y_teste, y_pred_base)

print("MODELO DE REFERÊNCIA, parâmetros padrão")
print("=" * 50)
print(f"Kernel: {svm_base.kernel}")
print(f"C: {svm_base.C}")
print(f"Gamma: {svm_base.gamma}")
print()
print(f"Acurácia no conjunto de teste: {acc_base*100:.2f}%")
```

## 17. Grid Search

```python
grade = {
    "C": [0.1, 1, 10, 100],
    "gamma": [0.001, 0.01, 0.1, 1, "scale"],
    "kernel": ["rbf"]
}

total = len(grade["C"]) * len(grade["gamma"]) * len(grade["kernel"])
print(f"Combinações a avaliar: {total}")
print(f"Validação cruzada: 5 partições")
print(f"Total de treinamentos: {total * 5}")
```

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

grid = GridSearchCV(
    estimator=SVC(random_state=42),
    param_grid=grade,
    cv=cv,
    scoring="accuracy",
    n_jobs=-1,
    verbose=1
)

grid.fit(X_treino_norm, y_treino)

print("\nBusca concluída.")
```

```python
print("RESULTADO DA BUSCA EM GRADE")
print("=" * 50)
print("Melhores hiperparâmetros encontrados:")
for parametro, valor in grid.best_params_.items():
    print(f"  {parametro}: {valor}")
print()
print(f"Acurácia média na validação cruzada: {grid.best_score_:.4f}, ou {grid.best_score_*100:.2f}%")
```

## 18. Análise dos resultados

```python
resultados = pd.DataFrame(grid.cv_results_)
colunas = ["param_C", "param_gamma", "param_kernel",
           "mean_test_score", "std_test_score", "rank_test_score"]

top10 = resultados[colunas].sort_values("rank_test_score").head(10)
top10.reset_index(drop=True)
```

```python
rbf = resultados[resultados["param_kernel"] == "rbf"].copy()
rbf["param_gamma"] = rbf["param_gamma"].astype(str)

pivot = rbf.pivot_table(index="param_C", columns="param_gamma",
                        values="mean_test_score")

plt.figure(figsize=(10, 6))
sns.heatmap(pivot, annot=True, fmt=".4f", cmap="YlGnBu", linewidths=0.5)
plt.title("Acurácia média na validação cruzada, kernel RBF", fontsize=13)
plt.xlabel("gamma")
plt.ylabel("C")
plt.tight_layout()
plt.show()
```

## 19. Treinamento do modelo final e avaliação no conjunto de teste

```python
modelo_final = grid.best_estimator_

y_pred = modelo_final.predict(X_teste_norm)

acuracia = accuracy_score(y_teste, y_pred)
precisao = precision_score(y_teste, y_pred)
revocacao = recall_score(y_teste, y_pred)
f1 = f1_score(y_teste, y_pred)

print("DESEMPENHO NO CONJUNTO DE TESTE")
print("=" * 50)
print(f"Acurácia:  {acuracia*100:.2f}%")
print(f"Precisão:  {precisao:.4f}")
print(f"Revocação: {revocacao:.4f}")
print(f"F1-score:  {f1:.4f}")
```

```python
print("RELATÓRIO DE CLASSIFICAÇÃO\n")
print(classification_report(y_teste, y_pred,
                            target_names=["Não potável", "Potável"],
                            digits=4))
```
